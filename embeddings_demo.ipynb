# Initial visualization of the lemma

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
df = pd.read_csv(inputfilename)
df.head()
```

```python
# compute embeddings for the target lemma
# in each sentence context
embeddings = compute_embeddings_for(df, what = "tgt", modeltype = modeltype)
```

```python
%matplotlib widget

contextatlas(df, embeddings = embeddings, numclusters = 7);
```

# Clustering

This visualization implicitly uses clustering (k-means) to group the data to compute indicative words. We can make the clusters visible:

```python
clusterlabels, cluster_maxword = contextatlas(df, embeddings = embeddings, numclusters = 7, clustercolors = True, showwords = False)
```

We can look at the clusters in more detail:

```python
sentence_snippets = make_sentence_snippets(df, windowsize = 10)
for ix, word in cluster_maxword.items():
    print(ix, word)
    for s in [s for i, s in zip(clusterlabels, sentence_snippets) if i == ix][:2]:
        print(s)
    print()
```
